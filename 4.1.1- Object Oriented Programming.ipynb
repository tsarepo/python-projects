## Introduction

There are different styles of programming - which also have the fancy name "programming paradigms". (Paradigm is such a cool word, that was made popular by the famous philosopher of Science Thomas Kuhn and his concept of "paradigm shift". Check him out: https://en.wikipedia.org/wiki/Thomas_Kuhn).

A programming paradigm does not refer to a specific language but rather to a way to program, a methodology or structuring. Some programming languages allow specific programming paradigms and not others.

Until now we have been using the *procedural* programming paradigm, also known as "inline programming". That is, a way to code our program with functions, variables, and by putting all our code in one file with a sequence of instructions, all executed one by one - sequentially. That’s the most basic way to code.

Python (as well as other programming languages) offer the possibility to code in a fancier way: *Object Oriented Programming*. This programming paradigm is based on the concept of "objects", which may contain data, in the form of "attributes" or "properties" as well as procedural code, known as "methods".

An OOP application will use a collection of objects which know how to perform certain actions and how to interact with other elements of the application. For example an object could be a person. That person would have a name (that would be a property of the object), and would know how to walk (that would be a method). A method in OOP can be considered as a procedure in PP, but here it belongs to a specific object. Another important aspect of OOP are classes. A class can be considered as a blueprint for an object.

More examples:

For instance, an object could represent a person with properties/attributes like a name, age, and address and behaviors/methods such as walking, talking, breathing, and running. Or it could represent an email with properties/attributes like a recipient list, subject, and body and behaviors/methods like adding attachments and sending. Now your turn: It could represent a car with properties attributes .... and behaviour/methods ....

Essentially, object-oriented programming is an approach for modeling concrete, real-world things, like cars, as well as relations between things, like companies and employees, students and teachers, and so on. OOP models real-world entities as software objects that have some data associated with them and can perform certain functions.

## Class vs Instance

Classes are used to create user-defined data structures. Classes define functions called methods, which identify the behaviors and actions that an object created from the class can perform with its data.

A class is a blueprint for how something should be defined. It doesn’t actually contain any data. The Employee class specifies that a name and an age are necessary for defining an employee, but it doesn’t contain the name or age of any specific employee.

While the class is the blueprint, an instance is an object that is built from a class and contains real data. An instance of the Employee class is not a blueprint anymore. It’s an actual employee with a name, like jack, who’s 30 years old.

Put another way, a class is like a form or __questionnaire__. An instance is like a __form that has been filled out with information__. Just like many people can fill out the same form with their own unique information, many instances can be created from a single class.

Perhaps you start to see what's the advantage of OOP a bit better: imagine you had to create a software for onboarding new employees to a company and registering their details - how would you do it?

To conclude on the topic of class vs instance, look at this code example:

## Class definition

```python
class Employee:
    def __init__(self, name, age, job):
        self.name = name
        self.age = age
        self.job = job
```

Ok what happened there?

1. We defined a class named Employee. It kinda looks a bit like the way we deal with functions with the keyword "def". Here the keyword is "class".
2. We defined the __ init __ method, this time with the "def" keyword that we use for functions. This is a method, so it makes sense intuitively to think about it in a similar way that you'd think of a function.
3. Why are we using __ init __ written like this, and what does it mean? This is sometimes called a "constructor", and essentially is meant initialize the class's attribute, and no other purpose than that. So here, we want to initialize the employee's name, age and job.
4. We are passing 3 arguments to the __ init __ constructor. These are the the parameters we want to be using, exactly like how we typically do it with a function. We want our Employee's name to be like the variable name that we passed, the Employee's age to be like the variable age, and so on.
5. What's up with "self"? It is used to represent the instance of the class - this is how we access attributes and methods that are internal to the class. Here we essentially created attributes (name, age and job) and by writing them after self., we specify that we're tying to create a specific instance of the class which will have name, age and job following the ones we pass as arguments to our __ init __ method.

Here are some hopefully self explanatory bits of code that follow from this class definition in order for us to us it!

## Instance methods

Instance methods are functions that are defined inside a class and can only be called from an instance of that class. Just like .__init__(), an instance method’s first parameter is always self.

```python
class Dog:
    species = "Canis familiaris"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Instance method
    def dog_description(self):
        return str(self.name + ' is ' + str(self.age) + ' years old')

    # Another instance method

    ## Notice how we don't need "self" for the sound variable.
    ## It's not an attribute of the class, just something it might
    ## our dog might say at some point!
    def speak(self, sound):
        return str(self.name + ' says ' + sound)

```

```python
Bobby = Dog("bobby", 4) ##Creating the instance

print(Bobby.dog_description()) ##Calling methods of that class instance to do stuff

print(Bobby.speak("Woof Woof")) ## Same
print(Bobby.speak("Woooooooooooooof plz give me food")) ## Same again

```
