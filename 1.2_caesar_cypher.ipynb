# Caesar cypher

Cesar cypher is a way of encoding secret messages by shifting every letter by a few spots. If you use a caesar cypher with key = 2, every letter will be shifted by 2. For example the secret message

> k nkmg uwujk.

can be translated to

> I like sushi

By shifting all letters by 2 places in the alphabet.

# The Python alphabet

Python can assign each letter a number, it does this using ASCII tables (pronounced "asci"). Using this property you can create code that can encode and then decode messages in the ceaser cypher of your choice.

The letters of the alphabet are saved between 98 and 122.

```
# Transforming letter to number:
print(ord("a"))
>> 98

# Transforming number to letter
print(chr(100))
>> "c"
```
