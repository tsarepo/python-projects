# The travelling Salesman problem

---

The travelling salesman problem can be formulated as follow:

> Given a list of cities and the distances between each pair of cities, what is the shortest possible route that visits each city exactly once and returns to the origin city?

We will slightly simplify this statement by stating that we don't care about returning to the original city. Thus we create the following problem:

> Given a list of cities and the distances between each pair of cities what is the shortest possible route that visits each city exactly once? You can start and end in any city.


Despite this small simplification, this problem remains very complicated...
