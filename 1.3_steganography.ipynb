### Representing images using bits and bytes

The value you will get for an individual pixel will be an integer between 0 and 255. This is how different shades on a black and white picture are represented.

How does a computer represent a number between 0 and 255? As we know, a computer only understands 0s and 1s. In order to represent the range 0-255, we need a combination of eight 0s and 1s to uniquely represent each number inside that range.

We call a single 0/1 a bit, and eight of them (i.e. enough to represent a number up to 255) we call a byte.

Inside a byte (= 8 zeros and ones), we can store a number.
For example, 11001000 = 200.
But inside a byte, we can also store a letter.
For example, 1000010 = B.

---

### So binary can store both visual and text information! This is at the heart of steganography, and the reason why we can even store text in an image in the first place.

Let’s look at which part of a byte tells us the most about the number it needs to encode.

10000000 = 128

10000001 = 129

00000001 = 1

The small difference between 128 and 129 is given by the number on the most right, but the large difference between 1 and 128 or 129 is given by the first number on the left. This translates into how easy it is to detect a change: the color difference between a pixel with value 128 and 129 will be impossible to detect by eye. We can therefore change these values without making it obvious from the outside where our “message” is hidden!

So, to summarize, what we will do in this project is finding and storing secret messages in the rightmost bits of a byte!
