### 1. Creating a dataset

```python
# Importing packages

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
# Setting a seed for reproducibility
np.random.seed(102)

# Generates random x values
x = np.linspace(0, 5, 100)

# Generates y
y = ((86/32)+0.58) * x + (0.32-4+5.46) + np.random.normal(0, 1, x.shape)

df = {'Tail wagging (Hz)': x, 'Happiness': y}


data = pd.DataFrame(df)


df2 = pd.DataFrame([[0, 32.34], [4.8, -3.5]], columns=['Tail wagging (Hz)', 'Happiness'], index=[100, 101])

data = pd.concat([data,df2])



# Generating second example

x = [i for i in range(1,11)]
y = [(2*i + 5 + np.random.normal(0,.6)) for i in x]
y[5] = 50

df = {'x': x, 'y': y}
example_data = pd.DataFrame(df)
```

### Outliers

If you look at the graph you'll notice that one of the values is extremely high and does not make sense. This can happen do to a typo while creating the data frame. We will first have to remove it before training our model because it may affect our performance negatively.
  

```python
### Data frame syntax ###

# (1) Finding row at which the variable is wrong:
row_id = np.where(example_data.y >= 30) # What does this function do here? If you can't figure it out, ask your teacher...
print(row_id)


# (2) Deleting that value in the data frame:
example_data = example_data.drop(5)

# (3) recreating x, y from corrected data:
x = example_data.x
y = example_data.y

```

We can now check if our data is clean by replotting it:

```python
plt.plot(x, y, 'ro') # r = red, o = circles to plot data
plt.xlabel("Variable x")
plt.ylabel("variable y")
plt.show
```

### Creating the model

The last thing left to do is creating, training, and testing our model. For this we use the sk-learn package. This package has many machine learning functions ranging from simple to insanely complicated stuff. We will be using the linear regression class (it's actually a class not a function). If you still aren't sure what a linear regression is: This is the time to ask your teacher!

```python
### Linear regression syntax ###

# Creating an instance of the linear regression class
Reg = LinearRegression()


# Acquiring data from dataframe
x = example_data.x
y = example_data.y

# Reshaping x to fit in training function
x = np.array(x)
x = x.reshape(-1 ,1)

# Training our instance on our data
Reg.fit(x, y)

# Extracting data from our regression:
print('Coefficient =', round(Reg.coef_[0], 3))
print('Intercept =', round(Reg.intercept_,3))

print()
print('Equation:')
print('y =', round(Reg.coef_[0], 3), '* x + ', round(Reg.intercept_,3))
```
